# Prepare Images for Labelling

Scans a folder for images that have a matching pipeline pickle and saves the
processed page image (`pre.bgr` — deskewed, cropped to content) into a
separate folder ready for import into a labelling tool.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import cv2
from pipeline.decorations import ManuscriptPage
```

```python
IMAGES_DIR = Path("../data/all_images")
OUTPUT_DIR = Path("../data/all_images_cropped")
```

```python
def find_pages(images_dir: Path) -> list[Path]:
    """Return image paths that have a matching pipeline pickle."""
    results_dir = images_dir / "results"
    pages = []
    for img in sorted(images_dir.glob("*.jpg")):
        pkl = results_dir / img.with_suffix(".pkl").name
        if pkl.is_file():
            pages.append(img)
        else:
            print(f"  [skip] no pickle for {img.name}")
    return pages
```

```python
image_paths = find_pages(IMAGES_DIR)
print(f"Found {len(image_paths)} pages")
```

```python
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

for img_path in image_paths:
    page = ManuscriptPage(img_path)
    out_path = OUTPUT_DIR / img_path.name
    cv2.imwrite(str(out_path), page.bgr)
    print(f"  saved {out_path.name}")

print(f"\nDone. {len(image_paths)} images saved to {OUTPUT_DIR.resolve()}")
```
