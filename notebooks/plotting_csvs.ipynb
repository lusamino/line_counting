```python
import pandas as pd
file = '/Users/luissalamanca/Dropbox/My_stuff/05_SDSCresearch/10_SideProjects/00_MedievalCambridge/line_counting/results/validation_counting_1st_allCols.csv'

df = pd.read_csv(file)
# Count all Trues values of columns: 'two_columns','interesting', 'recheck', 'hole_text', 'quire_mark', 'mix_layout','tail_rhyme'
interesting_true_count = df['interesting'].sum()
recheck_true_count = df['recheck'].sum()
two_columns_true_count = df['two_columns'].sum()
hole_text_true_count = df['hole_text'].sum()
quire_mark_true_count = df['quire_mark'].sum()
mix_layout_true_count = df['mix_layout'].sum()
tail_rhyme_true_count = df['tail_rhyme'].sum()

print(f"Interesting true count: {interesting_true_count}")
print(f"Recheck true count: {recheck_true_count}")
print(f"Two columns true count: {two_columns_true_count}")
print(f"Hole text true count: {hole_text_true_count}")
print(f"Quire mark true count: {quire_mark_true_count}")
print(f"Mix layout true count: {mix_layout_true_count}")
print(f"Tail rhyme true count: {tail_rhyme_true_count}")
```

```python
# Draw a plot with the page number in X and the number of lines in Y, and color the points based on the 'two_columns' column
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
# The page number is extracted from the filename, which has the format 'MS-GG-00001-00001-000-01274.jpg'
df['page_number'] = df['image'].apply(lambda x: int(x.split('-')[-1].split('.')[0]))
scatter = plt.scatter(df['page_number'], df['line_counting_validated'], c=df['two_columns'], cmap='coolwarm', alpha=0.7)
plt.colorbar(scatter, label='Two Columns')
plt.xlabel('Page Number')
plt.ylabel('Line Count')
plt.title('Line Count by Page Number (colored by Two Columns)')
plt.show()
```

```python
#####
### MARGING CSV

import pandas as pd
file = '/Users/luissalamanca/Dropbox/My_stuff/05_SDSCresearch/10_SideProjects/00_MedievalCambridge/line_counting/results/validation_counting_1st_allCols.csv'
file_visual = '/Users/luissalamanca/Dropbox/My_stuff/05_SDSCresearch/10_SideProjects/00_MedievalCambridge/line_counting/results/validation_decorations_allval_4categories_260608.csv'

file_merge = '/Users/luissalamanca/Dropbox/My_stuff/05_SDSCresearch/10_SideProjects/00_MedievalCambridge/line_counting/results/validation_lines_decorations.csv'
# Combine the two CSV files based on the 'image' column. Remove from the second csv the columns interesting and recheck
df_counting = pd.read_csv(file)
df_visual = pd.read_csv(file_visual)
df_visual = df_visual.drop(columns=['extracted_decorated_initial', 'extracted_large_decoration','extracted_paragraph_mark', 'extracted_miniatures','interesting', 'recheck'])

# Rename the elements in the image columns by replacing _decorations.json with .jpg
df_visual['image'] = df_visual['image'].apply(lambda x: x.replace('_decorations.json', '.jpg'))

df_merged = pd.merge(df_counting, df_visual, on='image', how='inner')
df_merged.to_csv(file_merge, index=False)
```
